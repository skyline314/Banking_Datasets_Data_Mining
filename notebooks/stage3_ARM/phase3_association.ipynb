# Stage 3: Association Rule Mining (ARM)

This notebook demonstrates the Association Rule Mining pipeline, which aims to discover frequent itemsets and interesting co-occurrence patterns in customer transactions.

```python
import os
import sys
import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# Add project root to path
project_root = str(Path(os.getcwd()).parent.parent)
if project_root not in sys.path:
    sys.path.insert(0, project_root)

# Import ARM modules
from src.stage3_ARM.discretize import load_and_discretize
from src.stage3_ARM.mine import run_mining
from src.stage3_ARM.interpret import interpret_and_export
from config.config import CATEGORICAL_DATA_PATH, ASSOC_OUTPUT_DIR
```

## 1. Load and Discretize Data

First, we load the pre-cleaned categorical dataset (`clean_categorical.csv`) and discretize continuous variables into domain-meaningful categorical bins (e.g., Balance: Very Low, Age: 25-34).

```python
df_discretized, transactions = load_and_discretize(CATEGORICAL_DATA_PATH)

display(df_discretized.head())
print(f"\nBuilt {len(transactions):,} transaction baskets for Apriori.")
print(f"Sample basket: {transactions[0]}")
```

## 2. Apriori Mining & Rule Generation

Next, we run the Apriori algorithm to discover frequent itemsets, then generate association rules. We filter for rules with high confidence and positive lift (Lift > 1).

```python
frequent_itemsets, rules, raw_rule_count = run_mining(transactions)

print(f"\nFrequent itemsets found: {len(frequent_itemsets):,}")
print(f"Rules before lift filter: {raw_rule_count:,}")
print(f"Rules after lift filter: {len(rules):,}")

display(rules.head(10))
```

## 3. Interpretation and Export

We interpret the discovered rules, generate plain-English explanations, and provide actionable business recommendations based on the patterns.

```python
top_rules = interpret_and_export(rules, frequent_itemsets, ASSOC_OUTPUT_DIR, raw_rule_count)

display(top_rules[["rule_text", "lift", "confidence", "support"]].head())
```
